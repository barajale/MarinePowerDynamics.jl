```julia
# import libraries
using Pkg
Pkg.activate("./env")
using ImportAll
@importall(MarinePowerDynamics)
include("./src/plotexample.jl")
include("./src/savedata.jl")
```

```julia
### Grab grid files
wec_powergrid = read_powergrid(joinpath(@__DIR__,"./src/WEC_IEEE14grid.json"), Json)

powergrid = read_powergrid(joinpath(@__DIR__,"./src/IEEE14grid.json"), Json)
```

### Flat run data

```julia
# Normal IEEE14 
grid = powergrid

operationpoint = find_operationpoint(grid)

# Configuration for the simulation
# notice the fault span is after the simulation start time to run a flat run
config = Dict(
    "fault_span" => (61, 62),
    "component" => "branch1",
    "timespan" => (0.0, 60.0),
    "description" => "IEEE14 flat run - no fault during simulation time",
    "grid_type" => "IEEE14_standard"
)
fault = LineFailure(line_name = config["component"], tspan_fault = config["fault_span"])

# run dynamic simulation
solution = simulate(fault, grid, operationpoint, config["timespan"])

# Save simulation results with metadata
file_info = save_sim(solution, config; 
                    output_dir="./simulation_results", 
                    filename_prefix="ieee14_flatrun",
                    variables=[:v, :p, :q, :ω])

grid = nothing # dereference grid 
```
